# Normalisation channel: $B^+ \to J/\psi(\mu^+\mu^-) K^+$

Extracts $N_{J/\psi K^+}$, the normalisation yield entering Eq. (3) of the report.

This channel is used instead of a hadronic normalisation mode so that the two branching fractions are measured independently rather than as a ratio — the $b \to c\bar{u}q$ anomaly affects both hadronic channels, so normalising one to the other would propagate any discrepancy into the result.

**Result:** $N_{J/\psi K^+} = 136\,071 \pm 428$, $\mu = 5281.49 \pm 0.06$ MeV/$c^2$.

## Candidate selection and mass histogram

Streams the HLT2-selected tuples, applies the requirements of Table 1 of the report (dimuon mass window, muon ID, track and vertex quality, flight-distance significance, bachelor `PIDK > 5`) and fills the $m(J/\psi K^+)$ histogram over 5100–5600 MeV/$c^2$.


```python
# Bachelor PIDK working points
BACHELOR_PIDK_MAX = 0   # Dπ: bachelor pion-like
BACHELOR_PIDK_MIN = 5   # DK: bachelor kaon-like

bachelor_pid_cut_Dpi = f"Kplus_PID_K < {BACHELOR_PIDK_MAX}"
bachelor_pid_cut_DK  = f"Kplus_PID_K > {BACHELOR_PIDK_MIN}"

# The bachelor here is a kaon, so the DK-like requirement applies.
bachelor_pid_cut = bachelor_pid_cut_DK

pid_cuts = [
    bachelor_pid_cut,
    "muplus_ISMUON && muminus_ISMUON",
    "muplus_PID_MU > 0 && muminus_PID_MU > 0",
]

non_pid_cuts = [
    "B_M>100 && B_M<5600",
    "B_BPVFDCHI2 > 100",
    "muplus_CHI2DOF < 10 && muminus_CHI2DOF < 10",
    "muplus_PT > 250 && muminus_PT > 250",
    "Jpsi_PT > 500",
    "abs(Jpsi_M-3096.9) < 80",
    "Jpsi_CHI2DOF < 20",
    "Kplus_PT > 250",
    "B_CHI2DOF < 20",
]

cuts = non_pid_cuts + pid_cuts
cut_str = " && ".join(cuts)

# Dataset in [5100, 5600] MeV after selection
mB  = ROOT.RooRealVar("B_M", "B_M [MeV/c^{2}]", 5100., 5600.)
rdf = ROOT.RDataFrame(decay_str, files).Filter(cut_str)

# Book a RooDataSet from the filtered dataframe
ds_ptr = rdf.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset","dataset", ROOT.RooArgSet(mB))),
    ("B_M",)
)
ds = ds_ptr.GetValue()
print(f"{int(ds.sumEntries())} entries in dataset after cuts (5100–5600 MeV)")

# Model: Signal (Crystal Ball if available, else Gaussian) + Chebychev(2), extended
use_crystal = hasattr(ROOT, "RooCrystalBall")

if use_crystal:
    mean   = ROOT.RooRealVar("mean",   "mean",   5279., 5265., 5290.)
    sigmaL = ROOT.RooRealVar("sigmaL", "sigmaL",   9.0,    3.,   30.)
    alphaL = ROOT.RooRealVar("alphaL", "alphaL",   1.5,   0.1,    5.)
    nL     = ROOT.RooRealVar("nL",     "nL",       5.0,   1.0,   50.)
    sigmaR = ROOT.RooRealVar("sigmaR", "sigmaR",   9.0,    3.,   30.)
    alphaR = ROOT.RooRealVar("alphaR", "alphaR",   2.0,   0.1,    5.)
    nR     = ROOT.RooRealVar("nR",     "nR",       5.0,   1.0,   50.)
    sig = ROOT.RooCrystalBall(
        "sig","Signal (CrystalBall)",
        mB, mean,
        sigmaL, alphaL, nL,
        sigmaR, alphaR, nR
    )
else:
    # Fallback: Gaussian
    mean  = ROOT.RooRealVar("mean",  "mean",  5279., 5265., 5290.)
    sigma = ROOT.RooRealVar("sigma", "sigma",   10.,    5.,   30.)
    sig   = ROOT.RooGaussian("sig", "Signal (Gauss)", mB, mean, sigma)

print("Using signal PDF:", sig.ClassName())

# Background: Chebychev(2)  (degree-2 polynomial)
c0 = ROOT.RooRealVar("c0","c0", 0.0, -1.0, 1.0)
c1 = ROOT.RooRealVar("c1","c1", 0.0, -1.0, 1.0)
c2 = ROOT.RooRealVar("c2","c2", 0.0, -1.0, 1.0)
bkg = ROOT.RooChebychev("bkg","Chebychev(2)", mB, ROOT.RooArgList(c0, c1, c2))

# Extended yields
nsig = ROOT.RooRealVar("nsig","signal yield",     5.0e4, 0.0, 1e9)
nbkg = ROOT.RooRealVar("nbkg","background yield", 5.0e5, 0.0, 1e10)

# Total model
model = ROOT.RooAddPdf("model","sig+bkg", ROOT.RooArgList(sig, bkg), ROOT.RooArgList(nsig, nbkg))

# Fit
fitres = model.fitTo(ds, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
fitres.Print()

# Plot: data, total, bkg-only, sig-only + legend & parameter box
COL_TOTAL = ROOT.kBlue+1
COL_BKG   = ROOT.kGreen+2
COL_SIG   = ROOT.kRed

c = ROOT.TCanvas("c_fit_all_cheb2","B_M fit: data + total + bkg + sig (Cheb-2)", 900, 700)
frame = mB.frame(ROOT.RooFit.Title("B^{+} #rightarrow J/#psi K^{+}  (Signal + Chebychev(2), with cuts)"),
                 ROOT.RooFit.Bins(60))

# Data
ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))

# Total model
model.plotOn(frame,
             ROOT.RooFit.LineColor(COL_TOTAL),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("model_curve"))

# Background component
model.plotOn(frame,
             ROOT.RooFit.Components("bkg"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(COL_BKG),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("bkg_curve"))

# Signal component
model.plotOn(frame,
             ROOT.RooFit.Components("sig"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(COL_SIG),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}}  [MeV/c^{2}]")
frame.GetYaxis().SetTitle("Events / bin")
frame.Draw()

# Legend
leg = ROOT.TLegend(0.58, 0.68, 0.90, 0.88)
leg.SetBorderSize(1); leg.SetFillColor(0)
leg.SetHeader("Components","C")
if frame.findObject("data_hist"):   leg.AddEntry(frame.findObject("data_hist"),   "Data",                  "lep")
if frame.findObject("model_curve"): leg.AddEntry(frame.findObject("model_curve"), "Total model",           "l")
if frame.findObject("bkg_curve"):   leg.AddEntry(frame.findObject("bkg_curve"),   "Background (Cheb-2)",  "l")
sig_label = "Signal (Crystal Ball)" if use_crystal else "Signal (Gaussian)"
if frame.findObject("sig_curve"):   leg.AddEntry(frame.findObject("sig_curve"),   sig_label,               "l")
leg.Draw()

# Parameter summary
box = ROOT.TPaveText(0.14, 0.70, 0.48, 0.88, "NDC")
box.SetFillStyle(0); box.SetBorderSize(0); box.SetTextFont(42); box.SetTextSize(0.035)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
if use_crystal:
    box.AddText(f"#sigma_{{L}} = {sigmaL.getVal():.2f} #pm {sigmaL.getError():.2f} MeV")
    box.AddText(f"#sigma_{{R}} = {sigmaR.getVal():.2f} #pm {sigmaR.getError():.2f} MeV")
else:
    box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.Draw()

c.Draw()
model_cb_cheb2 = model
ds_cb_cheb2 = ds
```

## Extended maximum-likelihood fit

Double-sided Crystal Ball signal plus a second-order Chebyshev combinatorial background. The Gaussian core represents the detector mass resolution; the asymmetric tails absorb final-state radiation from $J/\psi \to \mu^+\mu^-$ and residual non-Gaussian resolution.


```python
# Safety (recommended with RDataFrame + RooFit)
try:
    ROOT.ROOT.DisableImplicitMT()
except Exception:
    pass

# Common PID working points (USE THESE EVERYWHERE)
BACHELOR_PIDK_MAX = 0   # Dπ: bachelor pion-like
BACHELOR_PIDK_MIN = 5   # DK: bachelor kaon-like

bachelor_pid_cut_Dpi = f"Kplus_PID_K < {BACHELOR_PIDK_MAX}"
bachelor_pid_cut_DK  = f"Kplus_PID_K > {BACHELOR_PIDK_MIN}"
bachelor_pid_cut = bachelor_pid_cut_DK  # control channel: K-like

pid_cuts = [
    bachelor_pid_cut,
    "muplus_ISMUON && muminus_ISMUON",
    "muplus_PID_MU > 0 && muminus_PID_MU > 0",
]

# Mass windows
B_FULL_LO, B_FULL_HI = 5100.0, 5600.0
B_LO, B_HI           = 5150.0, 5600.0   # tight fit window

# Cuts for FULL dataset (for shape determination)
non_pid_cuts_full = [
    f"B_M>{B_FULL_LO} && B_M<{B_FULL_HI}",
    "B_BPVFDCHI2 > 100",
    "muplus_CHI2DOF < 10 && muminus_CHI2DOF < 10",
    "muplus_PT > 250 && muminus_PT > 250",
    "Jpsi_PT > 500",
    "abs(Jpsi_M-3096.9) < 80",
    "Jpsi_CHI2DOF < 20",
    "Kplus_PT > 250",
    "B_CHI2DOF < 20",
]
cut_str_full = " && ".join(non_pid_cuts_full + pid_cuts)

print("[PID consistency check]")
print("  bachelor PID cut:", bachelor_pid_cut)
print("\nFULL cut_str:\n", cut_str_full, "\n")

# Build ONE dataset (FULL), then derive tight dataset by reduce
mB = ROOT.RooRealVar("B_M", "B_M [MeV/c^{2}]", B_FULL_LO, B_FULL_HI)

rdf_full = ROOT.RDataFrame(decay_str, files).Filter(cut_str_full)
ds_ptr_full = rdf_full.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset_full", "dataset_full", ROOT.RooArgSet(mB))),
    ("B_M",)
)
ds_full = ds_ptr_full.GetValue()
print(f"{int(ds_full.sumEntries())} entries in FULL dataset ({B_FULL_LO:.0f}–{B_FULL_HI:.0f})")

# Tight dataset (same selection, just mass window)
ds_tight = ds_full.reduce(f"B_M>{B_LO} && B_M<{B_HI}")
print(f"{int(ds_tight.sumEntries())} entries in TIGHT dataset ({B_LO:.0f}–{B_HI:.0f})")

# 1) FULL fit: determine SIGNAL SHAPE (and background too)
use_crystal = hasattr(ROOT, "RooCrystalBall")

if use_crystal:
    mean   = ROOT.RooRealVar("mean",   "mean",   5279., 5265., 5290.)
    sigmaL = ROOT.RooRealVar("sigmaL", "sigmaL",   9.0,    3.,   30.)
    alphaL = ROOT.RooRealVar("alphaL", "alphaL",   1.5,   0.1,    5.)
    nL     = ROOT.RooRealVar("nL",     "nL",       5.0,   1.0,   50.)
    sigmaR = ROOT.RooRealVar("sigmaR", "sigmaR",   9.0,    3.,   30.)
    alphaR = ROOT.RooRealVar("alphaR", "alphaR",   2.0,   0.1,    5.)
    nR     = ROOT.RooRealVar("nR",     "nR",       5.0,   1.0,   50.)
    sig = ROOT.RooCrystalBall(
        "sig","Signal (CrystalBall)",
        mB, mean,
        sigmaL, alphaL, nL,
        sigmaR, alphaR, nR
    )
    sig_shape_params = [mean, sigmaL, alphaL, nL, sigmaR, alphaR, nR]
else:
    mean  = ROOT.RooRealVar("mean",  "mean",  5279., 5265., 5290.)
    sigma = ROOT.RooRealVar("sigma", "sigma",   10.,    5.,   30.)
    sig   = ROOT.RooGaussian("sig","Signal (Gauss)", mB, mean, sigma)
    sig_shape_params = [mean, sigma]

c0 = ROOT.RooRealVar("c0","c0", 0.0, -1.0, 1.0)
c1 = ROOT.RooRealVar("c1","c1", 0.0, -1.0, 1.0)
c2 = ROOT.RooRealVar("c2","c2", 0.0, -1.0, 1.0)
bkg = ROOT.RooChebychev("bkg","Chebychev(2)", mB, ROOT.RooArgList(c0, c1, c2))

nsig = ROOT.RooRealVar("nsig","signal yield",     5.0e4, 0.0, 1e9)
nbkg = ROOT.RooRealVar("nbkg","background yield", 5.0e5, 0.0, 1e10)

model = ROOT.RooAddPdf("model","sig+bkg", ROOT.RooArgList(sig, bkg), ROOT.RooArgList(nsig, nbkg))

fitres_full = model.fitTo(ds_full, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
print("FULL fit status =", fitres_full.status(), "covQual =", fitres_full.covQual())

# Freeze SIGNAL SHAPE (THIS IS THE KEY STEP FOR OPTION B)
#   - Keep background shape and yields floating in tight fit
for v in sig_shape_params:
    v.setConstant(True)

# 2) TIGHT fit: fit only yields (+background coeffs) with fixed signal shape
# define a named range and use Range(...) so Chebychev mapping stays consistent.
mB.setRange("tight", B_LO, B_HI)

fitres_tight = model.fitTo(
    ds_tight,
    ROOT.RooFit.Range("tight"),
    ROOT.RooFit.Extended(True),
    ROOT.RooFit.Save(True),
    ROOT.RooFit.PrintLevel(1),
    ROOT.RooFit.Strategy(2),
    ROOT.RooFit.Hesse(True),
    ROOT.RooFit.Offset(True),
)
print("TIGHT fit status =", fitres_tight.status(), "covQual =", fitres_tight.covQual())
```

## Validation: pion mass hypothesis

The bachelor track is reconstructed under the pion mass hypothesis to probe contamination from $B^+ \to J/\psi \pi^+$. No peaking structure is expected inside the kaon signal region.


```python
decay_str = "Hlt2B2CC_BuToJpsiKplus_JpsiToMuMu_Detached/DecayTree"

# Tight window (MeV)
B_LO, B_HI = 5150.0, 5600.0

# No PID cuts (quality/kinematics only) — IMPORTANT: no B_M window here
pre_cuts = [
    "B_BPVFDCHI2 > 100",
    "muplus_ISMUON && muminus_ISMUON",
    "muplus_CHI2DOF < 10 && muminus_CHI2DOF < 10",
    "muplus_PT > 250 && muminus_PT > 250",
    "Jpsi_PT > 500",
    "abs(Jpsi_M-3096.9) < 80",
    "Jpsi_CHI2DOF < 20",
    "Kplus_PT > 250",
    "B_CHI2DOF < 20",
]
pre_cut = " && ".join(pre_cuts)

# Helper C++ (MeV everywhere) with header guard to avoid redefinition
ROOT.gInterpreter.Declare(r"""
#ifndef E_FROM_P_MEV_DEFINED
#define E_FROM_P_MEV_DEFINED
#include <cmath>
#include <Math/Vector4D.h>
using V4 = ROOT::Math::PxPyPzEVector;
// px,py,pz in MeV; mMeV in MeV; return E in MeV
inline double E_from_p_MeV(double px, double py, double pz, double mMeV){
  return std::sqrt(px*px + py*py + pz*pz + mMeV*mMeV);
}
#endif
""")

# PDG masses (MeV)
m_mu = 105.6583745
m_pi = 139.57039
m_K  = 493.677

# Recompute masses FIRST (MeV!)
rdf0 = (ROOT.RDataFrame(decay_str, files)
        .Filter(pre_cut)
        # muon energies and 4-vectors (MeV)
        .Define("muplus_E",   f"E_from_p_MeV(muplus_PX,  muplus_PY,  muplus_PZ,  {m_mu})")
        .Define("muminus_E",  f"E_from_p_MeV(muminus_PX, muminus_PY, muminus_PZ, {m_mu})")
        .Define("lv_muplus",  "V4(muplus_PX,  muplus_PY,  muplus_PZ,  muplus_E)")
        .Define("lv_muminus", "V4(muminus_PX, muminus_PY, muminus_PZ, muminus_E)")
        .Define("lv_Jpsi",    "lv_muplus + lv_muminus")
        # bachelor energies (MeV) under K and π hypotheses
        .Define("E_bach_K",   f"E_from_p_MeV(Kplus_PX, Kplus_PY, Kplus_PZ, {m_K})")
        .Define("E_bach_pi",  f"E_from_p_MeV(Kplus_PX, Kplus_PY, Kplus_PZ, {m_pi})")
        .Define("lv_bach_K",  "V4(Kplus_PX, Kplus_PY, Kplus_PZ, E_bach_K)")
        .Define("lv_bach_pi", "V4(Kplus_PX, Kplus_PY, Kplus_PZ, E_bach_pi)")
        # B candidate masses (MeV) — NO *1000
        .Define("B_M_Khypo",  "(lv_Jpsi + lv_bach_K ).M()")
        .Define("B_M_Pihypo", "(lv_Jpsi + lv_bach_pi).M()")
)

print("entries before mass window:", rdf0.Count().GetValue())

# Apply tight window ON THE RECOMPUTED VARIABLES
rdfK  = rdf0.Filter(f"B_M_Khypo  > {B_LO} && B_M_Khypo  < {B_HI}")
rdfPi = rdf0.Filter(f"B_M_Pihypo > {B_LO} && B_M_Pihypo < {B_HI}")
print("entries in window (K-hypo):",  rdfK.Count().GetValue())
print("entries in window (π-hypo):",  rdfPi.Count().GetValue())

# Quick sanity hist (optional)
hK  = rdfK.Histo1D(("hK","K-hypo; m [MeV]; Events",60,B_LO,B_HI), "B_M_Khypo")
hPi = rdfPi.Histo1D(("hPi","π-hypo; m [MeV]; Events",60,B_LO,B_HI), "B_M_Pihypo")

# RooFit datasets: use an observable NOT present in the tree
mB = ROOT.RooRealVar("B_M_plot","m_{J/#psi+bachelor} [MeV/c^{2}]", B_LO, B_HI)
dsK  = rdfK .Book(ROOT.std.move(ROOT.RooDataSetHelper("dsK","dsK",   ROOT.RooArgSet(mB))), ("B_M_Khypo",)).GetValue()
dsPi = rdfPi.Book(ROOT.std.move(ROOT.RooDataSetHelper("dsPi","dsPi", ROOT.RooArgSet(mB))), ("B_M_Pihypo",)).GetValue()
print(f"dsK entries: {int(dsK.sumEntries())} | dsPi entries: {int(dsPi.sumEntries())}")

# Models: CB+Cheb(2) for K; Cheb(2) for π
have_cb = hasattr(ROOT,"RooCBShape")
if have_cb:
    mean  = ROOT.RooRealVar("mean","",5280.,5260.,5300.)
    sigma = ROOT.RooRealVar("sigma","",  12.,   4.,  40.)
    alpha = ROOT.RooRealVar("alpha","", 1.8,  0.5,   8.)
    n     = ROOT.RooRealVar("n","",       8.,  1., 100.)
    sig   = ROOT.RooCBShape("sig","Signal", mB, mean, sigma, alpha, n)
else:
    mean   = ROOT.RooRealVar("mean","",5280.,5260.,5300.)
    sigmaL = ROOT.RooRealVar("sigmaL","",12.,4.,40.)
    alphaL = ROOT.RooRealVar("alphaL","",1.8,0.5,8.0)
    nL     = ROOT.RooRealVar("nL","",8.,1.,100.)
    alphaR = ROOT.RooRealVar("alphaR","",10.0); alphaR.setConstant(True)
    nR     = ROOT.RooRealVar("nR","",50.0);     nR.setConstant(True)
    sig    = ROOT.RooCrystalBall("sig","Signal", mB, mean, sigmaL, alphaL, nL, sigmaL, alphaR, nR)

# Backgrounds
c0K = ROOT.RooRealVar("c0K","",0.,-1.,1.); c1K = ROOT.RooRealVar("c1K","",0.,-1.,1.); c2K = ROOT.RooRealVar("c2K","",0.,-1.,1.)
bkgK = ROOT.RooChebychev("bkgK","Cheb(2) K", mB, ROOT.RooArgList(c0K,c1K,c2K))

c0P = ROOT.RooRealVar("c0P","",0.,-1.,1.); c1P = ROOT.RooRealVar("c1P","",0.,-1.,1.); c2P = ROOT.RooRealVar("c2P","",0.,-1.,1.)
bkgPi = ROOT.RooChebychev("bkgPi","Cheb(2) π", mB, ROOT.RooArgList(c0P,c1P,c2P))

nsigK  = ROOT.RooRealVar("nsigK","", 1.2e5, 0., 1e11)
nbkgK  = ROOT.RooRealVar("nbkgK","", 3.0e5, 0., 1e12)
nbkgPi = ROOT.RooRealVar("nbkgPi","",3.0e5, 0., 1e12)

modelK  = ROOT.RooAddPdf("modelK","sig+bkg (K)", ROOT.RooArgList(sig,bkgK),  ROOT.RooArgList(nsigK,nbkgK))
modelPi = ROOT.RooAddPdf("modelPi","bkg only (π)", ROOT.RooArgList(bkgPi),   ROOT.RooArgList(nbkgPi))

fitK  = modelK.fitTo(dsK,  ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(1))
fitPi = modelPi.fitTo(dsPi, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(1))
print("fit statuses:", fitK.status(), fitPi.status())

# Overlay plot
COL_DATA_K, COL_MODEL_K  = ROOT.kBlack, ROOT.kRed+1
COL_DATA_PI, COL_MODEL_PI= ROOT.kBlue+2, ROOT.kAzure-6

can = ROOT.TCanvas("c_overlay_unitsOK","No PID: J/ψK vs J/ψπ (MeV-correct)",1000,760)
frame = mB.frame(ROOT.RooFit.Title("No PID: J/#psi K^{+} (CB+Cheb) vs J/#psi #pi (Cheb only)"),
                 ROOT.RooFit.Bins(60))

dsK.plotOn(frame, ROOT.RooFit.Name("dataK"), ROOT.RooFit.MarkerStyle(20), ROOT.RooFit.MarkerColor(COL_DATA_K))
modelK.plotOn(frame, ROOT.RooFit.Name("modelK"), ROOT.RooFit.LineColor(COL_MODEL_K), ROOT.RooFit.LineWidth(3))
modelK.plotOn(frame, ROOT.RooFit.Components("bkgK"), ROOT.RooFit.Name("bkgK"),
              ROOT.RooFit.LineStyle(ROOT.kDotted), ROOT.RooFit.LineColor(COL_MODEL_K), ROOT.RooFit.LineWidth(3))

dsPi.plotOn(frame, ROOT.RooFit.Name("dataPi"), ROOT.RooFit.MarkerStyle(24), ROOT.RooFit.MarkerColor(COL_DATA_PI))
modelPi.plotOn(frame, ROOT.RooFit.Name("modelPi"), ROOT.RooFit.LineColor(COL_MODEL_PI), ROOT.RooFit.LineWidth(3))

frame.GetXaxis().SetTitle("m_{J/#psi + bachelor} [MeV/c^{2}]")
frame.GetYaxis().SetTitle("Events / bin")
frame.Draw()

leg = ROOT.TLegend(0.56,0.60,0.90,0.88)
for name,label,opt in [
    ("dataK","Data (K hypothesis)","lep"),
    ("modelK","Fit (sig+bkg, K)","l"),
    ("bkgK","Bkg (K)","l"),
    ("dataPi","Data (\u03C0 hypothesis)","lep"),
    ("modelPi","Fit (bkg, \u03C0)","l"),
]:
    o = frame.findObject(name)
    if o: leg.AddEntry(o,label,opt)
leg.Draw()

can.Draw()
```
