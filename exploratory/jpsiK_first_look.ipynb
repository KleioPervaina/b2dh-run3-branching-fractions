# Exercise 1 : See if you can make the B+ -> Jpsi(->mu+ mu-) K+ signal cleaner. Take a look at B_BPVFDCHI2 and Kplus_PIDK



```python
import ROOT
ROOT.ROOT.EnableImplicitMT()

ntuple = "Hlt2B2CC_BuToJpsiKplus_JpsiToMuMu_Detached/DecayTree"
df_allfiles = ROOT.RDataFrame(ntuple, files)

# ---------- helper: compute quick S/B metrics for given cuts ----------
# Signal window around the B+ peak; sidebands on both sides
sig_lo, sig_hi = 5250, 5310
sb1_lo, sb1_hi = 5100, 5200
sb2_lo, sb2_hi = 5400, 5600
sb_width = (sb1_hi - sb1_lo) + (sb2_hi - sb2_lo)
sig_width = (sig_hi - sig_lo)

def metrics_for(c_fdchi2, c_pidk):
    cut = (
        f"B_BPVFDCHI2 > {c_fdchi2} && "
        f"Kplus_PID_K > {c_pidk} && "
        "muplus_ISMUON && muminus_ISMUON && muplus_PID_MU>0 && muminus_PID_MU>0"
    )
    d = df_allfiles.Filter(cut)
    S = d.Filter(f"{sig_lo}<B_M && B_M<{sig_hi}").Count().GetValue()
    B = d.Filter(f"({sb1_lo}<B_M && B_M<{sb1_hi}) || ({sb2_lo}<B_M && B_M<{sb2_hi})").Count().GetValue()
    # estimate background under the signal window by scaling the sidebands
    B_in_sig = B * (sig_width / sb_width)
    S_over_B = (S / B_in_sig) if B_in_sig>0 else float("inf")
    signif = S / ((S + B_in_sig)**0.5) if (S + B_in_sig)>0 else 0.0
    N = d.Count().GetValue()
    return dict(fdchi2=c_fdchi2, pidk=c_pidk, S=S, Bside=B, B_in_sig=B_in_sig, S_over_B=S_over_B, signif=signif, N=N, cut=cut)

# ---------- 1) quick scan over a few reasonable thresholds ----------
fdchi2_grid = [50, 80, 100, 150, 200]
pidk_grid   = [0, 1, 2, 3, 4]

results = []
for fcut in fdchi2_grid:
    for pc in pidk_grid:
        results.append(metrics_for(fcut, pc))

# print a compact table sorted by S/√(S+B)
results_sorted = sorted(results, key=lambda r: r["signif"], reverse=True)
print(f"{'FDCHI2':>6} {'PIDK':>4} {'S':>8} {'Bside':>8} {'B@sig':>8} {'S/B':>8} {'S/sqrt(S+B)':>13} {'N':>9}")
for r in results_sorted[:10]:
    print(f"{r['fdchi2']:6} {r['pidk']:4} {r['S']:8d} {r['Bside']:8d} {int(r['B_in_sig']):8d} {r['S_over_B']:8.2f} {r['signif']:13.2f} {r['N']:9d}")

# Pick a working point you like (good balance of purity & stats)
best_fdchi2, best_pidk = 100, 2   # adjust if your scan suggests better values

# ---------- 2) plot B_M before vs after your chosen cuts ----------
c_ex1 = ROOT.TCanvas("c_ex1","Exercise 1: cleaner signal",1000,450)
c_ex1.Divide(2,1)

# no cuts
h_before = df_allfiles.Histo1D(("h_before",";B_{M} [MeV];Events / 10 MeV",50,5100,5600),"B_M")
c_ex1.cd(1); h_before.Draw()

# with cuts
sel = (
    f"B_BPVFDCHI2 > {best_fdchi2} && "
    f"Kplus_PID_K > {best_pidk} && "
    "muplus_ISMUON && muminus_ISMUON && muplus_PID_MU>0 && muminus_PID_MU>0"
)
df_sel_allfiles = df_allfiles.Filter(sel)
h_after = df_sel_allfiles.Histo1D(("h_after",f";B_{ { } }M [MeV];Events / 10 MeV",50,5100,5600),"B_M")
c_ex1.cd(2); h_after.Draw()
c_ex1.Draw()

# ---------- 3) print metrics for the chosen point ----------
final = metrics_for(best_fdchi2, best_pidk)
print("\nChosen working point:")
print(f"   Cut: {final['cut']}")
print(f"   Total events after cuts: {final['N']}")
print(f"   Signal window S: {final['S']}")
print(f"   Sidebands     B: {final['Bside']}  → est. B in sig win: {final['B_in_sig']:.1f}")
print(f"   S/B ≈ {final['S_over_B']:.2f}   ,   S/√(S+B) ≈ {final['signif']:.2f}")
```

```python
# Use the full dataset
ntuple_all = decay_str
df_all = ROOT.RDataFrame(ntuple_all, files)

# --- helper: build PxPyPzE four-vectors (unique name to avoid redefinition) ---
ROOT.gInterpreter.Declare(r"""
using V4 = ROOT::Math::PxPyPzEVector;
inline V4 makeP4_e2(double px,double py,double pz,double e){ return V4(px,py,pz,e); }
""")

# --- build mu+, mu-, K+ four-vectors; then J/psi and B+; then masses ---
df_masses = (
    df_all
    .Define("mup4_e2",   "makeP4_e2(muplus_FOURMOMENTUMX,  muplus_FOURMOMENTUMY,  muplus_FOURMOMENTUMZ,  muplus_FOURMOMENTUME)")
    .Define("mum4_e2",   "makeP4_e2(muminus_FOURMOMENTUMX, muminus_FOURMOMENTUMY, muminus_FOURMOMENTUMZ, muminus_FOURMOMENTUME)")
    .Define("kp4_e2",    "makeP4_e2(Kplus_FOURMOMENTUMX,   Kplus_FOURMOMENTUMY,   Kplus_FOURMOMENTUMZ,   Kplus_FOURMOMENTUME)")
    .Define("Jpsi_p4_e2","mup4_e2 + mum4_e2")
    .Define("B_p4_e2",   "Jpsi_p4_e2 + kp4_e2")
    .Define("Jpsi_M_new","Jpsi_p4_e2.M()")
    .Define("B_M_new",   "B_p4_e2.M()")
)

# --- Histograms (rebuilt vs stored), explicit binning ---
c_e2 = ROOT.TCanvas("c_e2","Exercise 2: mass reconstruction",1000,500)
c_e2.Divide(2,1)

# J/psi comparison
c_e2.cd(1)
hJ_new = df_masses.Histo1D(("hJ_new",";J/#psi mass [MeV];Events / 2 MeV",100,3000,3200),"Jpsi_M_new").GetValue()
hJ_sto = df_masses.Histo1D(("hJ_sto",";J/#psi mass [MeV];Events / 2 MeV",100,3000,3200),"Jpsi_M").GetValue()
# normalize to unit area for shape comparison
if hJ_new.Integral()>0: hJ_new.Scale(1.0/hJ_new.Integral())
if hJ_sto.Integral()>0: hJ_sto.Scale(1.0/hJ_sto.Integral())
hJ_new.SetLineWidth(2); hJ_sto.SetLineWidth(2); hJ_sto.SetLineColor(ROOT.kRed)
hJ_new.Draw("HIST"); hJ_sto.Draw("HIST SAME")
legJ = ROOT.TLegend(0.58,0.75,0.88,0.88); legJ.AddEntry(hJ_new,"rebuilt","l"); legJ.AddEntry(hJ_sto,"stored","l"); legJ.Draw()

# B+ comparison
c_e2.cd(2)
hB_new = df_masses.Histo1D(("hB_new",";B^{+} mass [MeV];Events / 10 MeV",50,5100,5600),"B_M_new").GetValue()
hB_sto = df_masses.Histo1D(("hB_sto",";B^{+} mass [MeV];Events / 10 MeV",50,5100,5600),"B_M").GetValue()
if hB_new.Integral()>0: hB_new.Scale(1.0/hB_new.Integral())
if hB_sto.Integral()>0: hB_sto.Scale(1.0/hB_sto.Integral())
hB_new.SetLineWidth(2); hB_sto.SetLineWidth(2); hB_sto.SetLineColor(ROOT.kRed)
hB_new.Draw("HIST"); hB_sto.Draw("HIST SAME")
legB = ROOT.TLegend(0.58,0.75,0.88,0.88); legB.AddEntry(hB_new,"rebuilt","l"); legB.AddEntry(hB_sto,"stored","l"); legB.Draw()

c_e2.Draw()

# --- Differences (rebuilt - stored) for a numerical check ---
hJ_diff = df_masses.Define("dJ_e2","Jpsi_M_new - Jpsi_M").Histo1D(("hJ_diff",";J/#psi (new - stored) [MeV];Events",80,-20,20),"dJ_e2").GetValue()
hB_diff = df_masses.Define("dB_e2","B_M_new - B_M").Histo1D(("hB_diff",";B^{+} (new - stored) [MeV];Events",80,-60,60),"dB_e2").GetValue()

print("⟨Jpsi_M_new - Jpsi_M⟩ = {:.3f} MeV,  RMS = {:.3f} MeV".format(hJ_diff.GetMean(), hJ_diff.GetStdDev()))
print("⟨B_M_new   - B_M⟩    = {:.3f} MeV,  RMS = {:.3f} MeV".format(hB_diff.GetMean(), hB_diff.GetStdDev()))
```

```python
import ROOT

# 0) MT before any RDataFrame
ROOT.ROOT.EnableImplicitMT()

# 1) Build a filtered dataframe over ALL files (adjust 'decay_str' and 'files' if needed)
decay_str = "Hlt2B2CC_BuToJpsiKplus_JpsiToMuMu_Detached/DecayTree"
cuts = "B_M>5100 && B_M<5600"  # keep fit range; avoids out-of-range entries
df_sel = ROOT.RDataFrame(decay_str, files).Filter(cuts)

# 2) Define RooFit observable (this is what 'lab0_MASS' was meant to be)
B_Mvar = ROOT.RooRealVar("B_M", "B_M", 5100., 5600.)

# 3) Book RooDataSet from the dataframe (uses the same variable)
ds_ptr = df_sel.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset", "dataset", ROOT.RooArgSet(B_Mvar))),
    ("B_M",)
)
ds = ds_ptr.GetValue()
print(f"{int(ds.sumEntries())} entries in dataset\n")

# 4) --- Signal: Gaussian ---
mean  = ROOT.RooRealVar("mean",  "mean of gaussians", 5275., 5250., 5285.)
sigma = ROOT.RooRealVar("sigma", "width of gaussians",   10.,    8.,   25.)
sig   = ROOT.RooGaussian("sig", "Signal", B_Mvar, mean, sigma)

# 5) --- Background: Chebychev (order 2) ---
c0 = ROOT.RooRealVar("c0","coefficient #0", -0.3, -1., 1.)
c1 = ROOT.RooRealVar("c1","coefficient #1",  0.1, -1., 1.)
bkg = ROOT.RooChebychev("bkg","background p.d.f.", B_Mvar, ROOT.RooArgList(c0,c1))

# 6) --- Extended yields ---
nsig = ROOT.RooRealVar("nsig","number of signal events",     3.3e4, 0., 1e9)
nbkg = ROOT.RooRealVar("nbkg","number of background events", 1.3e5, 0., 1e9)

# 7) Total model
model = ROOT.RooAddPdf("model","sig+bkg", ROOT.RooArgList(sig,bkg), ROOT.RooArgList(nsig,nbkg))

# 8) Fit
fitres = model.fitTo(ds, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
fitres.Print()

# 9) Plot + legend + parameter box
c = ROOT.TCanvas("extendedmlfit","extendedmlfit",700,600)
frame = B_Mvar.frame(ROOT.RooFit.Title("Extended ML Fit"), ROOT.RooFit.Bins(50))

ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))
model.plotOn(frame, ROOT.RooFit.Name("model_curve"))
model.plotOn(frame, ROOT.RooFit.Components("bkg"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(ROOT.kGreen+2),
             ROOT.RooFit.Name("bkg_curve"))
model.plotOn(frame, ROOT.RooFit.Components("sig"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(ROOT.kRed),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}} [MeV/c^{2}]")
frame.Draw()

leg = ROOT.TLegend(0.55, 0.68, 0.88, 0.88); leg.SetBorderSize(0); leg.SetFillStyle(0); leg.SetHeader("Components","C")
for name,label,opt in [("data_hist","Data","lep"),("model_curve","Total model","l"),
                       ("bkg_curve","Background (Chebychev)","l"),("sig_curve","Signal (Gaussian)","l")]:
    obj = frame.findObject(name)
    if obj: leg.AddEntry(obj, label, opt)
leg.Draw()

box = ROOT.TPaveText(0.15, 0.70, 0.48, 0.88, "NDC"); box.SetFillStyle(0); box.SetBorderSize(0)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.Draw()

c.Draw()
```

```python
import ROOT

# --- Use all files and enable multithreading BEFORE making any RDataFrame
ROOT.ROOT.EnableImplicitMT()
ntuple = "Hlt2B2CC_BuToJpsiKplus_JpsiToMuMu_Detached/DecayTree"

# --- Selection: SAME working point as before, with comments for each term
#   • B_M window: keep the fit range and prevent RooFit from skipping out-of-range entries
#   • B_BPVFDCHI2>100: significant B flight distance from the PV (detached B) → suppresses prompt/combinatorial bg
#   • Kplus_PID_K>2: kaon-likeness requirement → suppresses π mis-ID
#   • mu±_ISMUON: basic muon-ID requirement from the muon system
#   • mu±_PID_MU>0: positive muon PID likelihood → improves muon purity
cuts_with_comments = [
    "B_M > 5100 && B_M < 5600",                    # (fit window & avoid out-of-range events)
    "B_BPVFDCHI2 > 100",                           # (detachment significance: B flight-distance χ² from PV)
    "Kplus_PID_K > 2",                             # (kaon PID selection: reduces π→K mis-ID)
    "muplus_ISMUON && muminus_ISMUON",            # (both tracks identified as muons)
    "muplus_PID_MU > 0 && muminus_PID_MU > 0",    # (positive muon PID likelihood on both muons)
]
cut_str = " && ".join(cuts_with_comments)

# --- Build dataframe over ALL files and apply the cuts
df_sel = ROOT.RDataFrame(ntuple, files).Filter(cut_str)

# --- RooFit observable (same as before) over the same range as the mass window
B_M = ROOT.RooRealVar("B_M", "B_M", 5100., 5600.)

# --- Book a RooDataSet directly from the filtered dataframe
ds_ptr = df_sel.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset", "dataset", ROOT.RooArgSet(B_M))),
    ("B_M",)
)
ds = ds_ptr.GetValue()
print(f"{int(ds.sumEntries())} entries in dataset after cuts\n")

# ============================
# === SAME FIT AS BEFORE  ====
#   Signal: Gaussian
#   Background: Chebychev(2)
#   Extended fit for nsig, nbkg
# ============================

# Signal (Gaussian)
mean  = ROOT.RooRealVar("mean",  "mean of gaussians", 5275., 5250., 5285.)
sigma = ROOT.RooRealVar("sigma", "width of gaussians", 10.0,   8.0,  25.0)
sig   = ROOT.RooGaussian("sig", "Signal", B_M, mean, sigma)

# Background (Chebychev polynomial of order 2)
c0 = ROOT.RooRealVar("c0", "coefficient #0", -0.30, -1.0, 1.0)
c1 = ROOT.RooRealVar("c1", "coefficient #1",  0.10, -1.0, 1.0)
bkg = ROOT.RooChebychev("bkg", "background p.d.f.", B_M, ROOT.RooArgList(c0, c1))

# Extended yields (free parameters)
nsig = ROOT.RooRealVar("nsig", "number of signal events",     3.3e4, 0.0, 1e9)
nbkg = ROOT.RooRealVar("nbkg", "number of background events", 1.3e5, 0.0, 1e9)

# Total model = nsig*sig + nbkg*bkg
model = ROOT.RooAddPdf("model", "sig+bkg", ROOT.RooArgList(sig, bkg), ROOT.RooArgList(nsig, nbkg))

# --- Fit (extended maximum likelihood)
fitres = model.fitTo(ds, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
fitres.Print()

# --- Plot with a legend identifying each curve
c = ROOT.TCanvas("c_fit_wp", "Extended ML Fit with working-point cuts", 800, 650)
frame = B_M.frame(ROOT.RooFit.Title("B^{+} \\rightarrow J/#psi K^{+} (Gaussian + Chebychev), after cuts"), ROOT.RooFit.Bins(50))

# Name each drawable so we can put it in the legend
ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))
model.plotOn(frame, ROOT.RooFit.Name("model_curve"))
model.plotOn(frame, ROOT.RooFit.Components("bkg"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(ROOT.kGreen+2),
             ROOT.RooFit.Name("bkg_curve"))
model.plotOn(frame, ROOT.RooFit.Components("sig"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(ROOT.kRed),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}}  [MeV/c^{2}]")
frame.Draw()

# Legend (“table”) and a small parameter box
leg = ROOT.TLegend(0.58, 0.68, 0.88, 0.88); leg.SetBorderSize(0); leg.SetFillStyle(0); leg.SetHeader("Components","C")
for name, label, opt in [
    ("data_hist",   "Data",                 "lep"),
    ("model_curve", "Total model",          "l"),
    ("bkg_curve",   "Background (Chebychev)","l"),
    ("sig_curve",   "Signal (Gaussian)",    "l"),
]:
    obj = frame.findObject(name)
    if obj: leg.AddEntry(obj, label, opt)
leg.Draw()

box = ROOT.TPaveText(0.15, 0.72, 0.50, 0.88, "NDC"); box.SetFillStyle(0); box.SetBorderSize(0)
box.SetTextFont(42); box.SetTextSize(0.035)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.Draw()

c.Draw()
```

```python
# crystal ball and bft masss
import ROOT

# --- 0) Use all files and enable multithreading BEFORE making any RDataFrame
ROOT.ROOT.EnableImplicitMT()
ntuple = "Hlt2B2CC_BuToJpsiKplus_JpsiToMuMu_Detached/DecayTree"

# --- 1) Selection (same spirit as before), but cut on the DTF mass window we will fit
# Notes next to each cut explain the physics
cuts_with_comments = [
    "B_DTF_MASS > 5150 && B_DTF_MASS < 5550",     # (fit window in DTF mass; avoids out-of-range entries)
    "B_BPVFDCHI2 > 100",                           # (detachment significance: B flight-distance χ² from PV)
    "Kplus_PID_K > 2",                             # (kaon PID: reduces π→K mis-ID)
    "muplus_ISMUON && muminus_ISMUON",            # (both tracks identified as muons)
    "muplus_PID_MU > 0 && muminus_PID_MU > 0",    # (positive muon PID likelihood on both muons)
    # optional: tighten J/psi mass if available, helps purity even more
    # "abs(Jpsi_M - 3096.9) < 80",
]
cut_str = " && ".join(cuts_with_comments)

# --- 2) Build dataframe over ALL files and apply the cuts
df_sel = ROOT.RDataFrame(ntuple, files).Filter(cut_str)

# --- 3) RooFit observable: B_DTF_MASS, same range as the selection
Bdtf = ROOT.RooRealVar("B_DTF_MASS", "B_DTF_MASS", 5150., 5550.)

# --- 4) Book a RooDataSet directly from the filtered dataframe (one column: B_DTF_MASS)
ds_ptr = df_sel.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset", "dataset", ROOT.RooArgSet(Bdtf))),
    ("B_DTF_MASS",)
)
ds = ds_ptr.GetValue()
print(f"{int(ds.sumEntries())} entries in dataset after cuts\n")

# =======================================================
# === 5) Fit model: Signal = Crystal Ball, Bkg = Exp  ===
# =======================================================

# 5a) Crystal Ball parameters:
#   mean  = peak position
#   sigma = Gaussian core width
#   alpha = transition (in sigma units) to the power-law tail
#   n     = tail exponent; larger n = steeper tail
mean  = ROOT.RooRealVar("mean",  "CB mean",   5279., 5270., 5288.)
sigma = ROOT.RooRealVar("sigma", "CB sigma",     8.,    3.,   20.)
alpha = ROOT.RooRealVar("alpha", "CB alpha",   1.5,   0.5,    5.)
n     = ROOT.RooRealVar("n",     "CB n",       3.0,   1.0,   20.)

sig_cb = ROOT.RooCBShape("sig_cb", "Signal (Crystal Ball)", Bdtf, mean, sigma, alpha, n)

# 5b) Background = exponential in mass
tau = ROOT.RooRealVar("tau", "exp slope", -0.002, -1.0, 0.0)
bkg_exp = ROOT.RooExponential("bkg_exp", "Background (exp)", Bdtf, tau)

# 5c) Extended yields (free parameters)
nsig = ROOT.RooRealVar("nsig", "number of signal events",     3.3e4, 0.0, 1e9)
nbkg = ROOT.RooRealVar("nbkg", "number of background events", 1.3e5, 0.0, 1e9)

# 5d) Total model = nsig * sig_cb + nbkg * bkg_exp
model = ROOT.RooAddPdf("model", "sig+bkg", ROOT.RooArgList(sig_cb, bkg_exp), ROOT.RooArgList(nsig, nbkg))

# --- 6) (Optional but robust) Pre-fit background in sidebands to stabilize tau, then full fit
Bdtf.setRange("SB1", 5150., 5230.)
Bdtf.setRange("SB2", 5330., 5550.)
b_only = ROOT.RooAddPdf("b_only", "", ROOT.RooArgList(bkg_exp), ROOT.RooArgList(nbkg))
b_only.fitTo(ds, ROOT.RooFit.Range("SB1,SB2"), ROOT.RooFit.PrintLevel(-1))

# --- 7) Full extended ML fit
fitres = model.fitTo(ds, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
fitres.Print()

# --- 8) Plot
c = ROOT.TCanvas("c_fit_dtf", "B_DTF_MASS fit: CB + exp", 800, 650)
frame = Bdtf.frame(ROOT.RooFit.Title("B^{+} \\rightarrow J/#psi K^{+}  (Crystal Ball + Exponential) on B_{DTF}"), ROOT.RooFit.Bins(50))

ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))
model.plotOn(frame, ROOT.RooFit.Name("model_curve"))
model.plotOn(frame,
             ROOT.RooFit.Components("bkg_exp"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(ROOT.kGreen+2),
             ROOT.RooFit.Name("bkg_curve"))
model.plotOn(frame,
             ROOT.RooFit.Components("sig_cb"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(ROOT.kRed),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}}^{\\,DTF}  [MeV/c^{2}]")
frame.Draw()

# --- 9) Legend and a small parameter box
leg = ROOT.TLegend(0.58, 0.68, 0.88, 0.88); leg.SetBorderSize(0); leg.SetFillStyle(0); leg.SetHeader("Components","C")
for name, label, opt in [
    ("data_hist",   "Data",                 "lep"),
    ("model_curve", "Total model",          "l"),
    ("bkg_curve",   "Background (exp)",     "l"),
    ("sig_curve",   "Signal (Crystal Ball)","l"),
]:
    obj = frame.findObject(name)
    if obj: leg.AddEntry(obj, label, opt)
leg.Draw()

box = ROOT.TPaveText(0.15, 0.72, 0.50, 0.88, "NDC"); box.SetFillStyle(0); box.SetBorderSize(0)
box.SetTextFont(42); box.SetTextSize(0.035)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.AddText(f"#alpha = {alpha.getVal():.2f},  n = {n.getVal():.2f}")
box.AddText(f"#tau = {tau.getVal():.4g}")
box.Draw()

c.Draw()
```
