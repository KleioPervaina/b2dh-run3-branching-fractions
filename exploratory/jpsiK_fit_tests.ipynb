```python
# Quick efficiencies
rdf_all = ROOT.RDataFrame(decay_str, files)
n0 = rdf_all.Count().GetValue()

base = "B_M>5100 && B_M<5600"
loose = " && ".join([
    base,
    "B_BPVFDCHI2>100",
    "Kplus_PID_K>2",
    "muplus_ISMUON && muminus_ISMUON",
    "muplus_PID_MU>0 && muminus_PID_MU>0",
    "muplus_CHI2DOF<10 && muminus_CHI2DOF<10",
    "muplus_PT>250 && muminus_PT>250",
    "Jpsi_PT>500",
    "abs(Jpsi_M-3096.9)<80",
    "Jpsi_CHI2DOF<20",
    "Kplus_PT>250",
    "abs(B_M-5279.34)<300",
    "B_CHI2DOF<20",
])
nSel = rdf_all.Filter(loose).Count().GetValue()
print(f"Selected: {nSel:,} / {n0:,} = {100*nSel/max(1,n0):.1f}%")
```

```python
cuts = [
    "B_M>5100 && B_M<5600",
    "B_BPVFDCHI2 > 100",
    "Kplus_PID_K > 2",
    "muplus_ISMUON && muminus_ISMUON",
    "muplus_PID_MU > 0 && muminus_PID_MU > 0",
    "muplus_CHI2DOF < 10 && muminus_CHI2DOF < 10",
    "muplus_PT > 250 && muminus_PT > 250",
    "Jpsi_PT > 500",
    "abs(Jpsi_M-3096.9) < 80",
    "Jpsi_CHI2DOF < 20",
    "Kplus_PT > 250",
    "B_CHI2DOF < 20",
]

rdf0 = ROOT.RDataFrame(decay_str, files)
N0 = int(rdf0.Count().GetValue())

cum = []
expr = ""
lastN = N0
for c in cuts:
    expr = c if not expr else f"({expr}) && ({c})"
    N = int(rdf0.Filter(expr).Count().GetValue())
    cum.append((c, N, N/lastN if lastN>0 else 0.0, N/N0 if N0>0 else 0.0))
    lastN = N

print("Cut                                    | kept | step eff | cum eff")
for c,N,step,cumEff in cum:
    print(f"{c:38} | {N:7d} | {step:8.3f} | {cumEff:7.3f}")
```

```python
# windows (MeV)
sigL, sigH   = 5250, 5310
side1 = (5100, 5200)
side2 = (5400, 5600)
rdf_before = ROOT.RDataFrame(decay_str, files).Filter("B_M>5100 && B_M<5600")
rdf_after  = ROOT.RDataFrame(decay_str, files).Filter(" && ".join(cuts))

def counts(rdf):
    S   = int(rdf.Filter(f"B_M>{sigL} && B_M<{sigH}").Count().GetValue())
    B1  = int(rdf.Filter(f"B_M>{side1[0]} && B_M<{side1[1]}").Count().GetValue())
    B2  = int(rdf.Filter(f"B_M>{side2[0]} && B_M<{side2[1]}").Count().GetValue())
    # scale sidebands into signal window width
    wSig = sigH - sigL
    wB   = (side1[1]-side1[0]) + (side2[1]-side2[0])
    Best = (B1+B2) * (wSig / wB)
    return S, Best

S0, B0 = counts(rdf_before)
S1, B1 = counts(rdf_after)
def fmt(S,B):
    import math
    return f"S={S:,},  B≈{B:,.0f},  S/B≈{S/max(1,B):.2f},  S/√(S+B)≈{S/max(1, (S+B))**0.5:.1f}"

print("Before:", fmt(S0,B0))
print("After: ", fmt(S1,B1))
```

### Double Crystal Ball + Chebychev(1)


```python
# new cuts
cuts = [
    "B_M>5100 && B_M<5600",                     # fit/plot window only
    "B_BPVFDCHI2 > 100",                        # detached B (flight-distance chi2)
    "Kplus_PID_K > 2",                          # kaon PID
    "muplus_ISMUON && muminus_ISMUON",          # both are muons
    "muplus_PID_MU > 0 && muminus_PID_MU > 0",  # positive muon PID likelihood
    "muplus_CHI2DOF < 10 && muminus_CHI2DOF < 10",  # reasonable track fits
    "muplus_PT > 250 && muminus_PT > 250",      # loose pT for muons (MeV)
    "Jpsi_PT > 500",                             # J/ψ pT (MeV)
    "abs(Jpsi_M-3096.9) < 80",                  # J/ψ mass window (MeV)
    "Jpsi_CHI2DOF < 20",                        # good J/ψ vertex fit
    "Kplus_PT > 250",                           # kaon pT (MeV)
    "B_CHI2DOF < 20",                           # decent B vertex fit
]
cut_str = " && ".join(cuts)
# Dataset in [5100,5600] MeV WITH CUTS
mB  = ROOT.RooRealVar("B_M", "B_M [MeV/c^{2}]", 5100., 5600.)
rdf = ROOT.RDataFrame(decay_str, files).Filter(cut_str)

# Book a RooDataSet from the filtered dataframe
ds_ptr = rdf.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset","dataset", ROOT.RooArgSet(mB))),
    ("B_M",)
)
ds = ds_ptr.GetValue()
print(f"{int(ds.sumEntries())} entries in dataset after cuts (5100–5600 MeV)")

# Model: Signal (Crystal Ball if available, else Gaussian) + Chebychev(1), extended
use_crystal = hasattr(ROOT, "RooCrystalBall")

if use_crystal:
    mean   = ROOT.RooRealVar("mean",   "mean",   5279., 5265., 5290.)
    sigmaL = ROOT.RooRealVar("sigmaL", "sigmaL",   9.0,    3.,   30.)
    alphaL = ROOT.RooRealVar("alphaL", "alphaL",   1.5,   0.1,    5.)
    nL     = ROOT.RooRealVar("nL",     "nL",       5.0,   1.0,   50.)
    sigmaR = ROOT.RooRealVar("sigmaR", "sigmaR",   9.0,    3.,   30.)
    alphaR = ROOT.RooRealVar("alphaR", "alphaR",   2.0,   0.1,    5.)
    nR     = ROOT.RooRealVar("nR",     "nR",       5.0,   1.0,   50.)
    sig = ROOT.RooCrystalBall(
        "sig","Signal (CrystalBall)",
        mB, mean,
        sigmaL, alphaL, nL,
        sigmaR, alphaR, nR
    )
else:
    # Fallback
    mean  = ROOT.RooRealVar("mean",  "mean",  5279., 5265., 5290.)
    sigma = ROOT.RooRealVar("sigma", "sigma",   10.,    5.,   30.)
    sig   = ROOT.RooGaussian("sig", "Signal (Gauss)", mB, mean, sigma)

print("Using signal PDF:", sig.ClassName())

# Background: Chebychev(1) (two coeffs -> first order)
c0 = ROOT.RooRealVar("c0","c0", 0.0, -1.0, 1.0)
c1 = ROOT.RooRealVar("c1","c1", 0.0, -1.0, 1.0)
bkg = ROOT.RooChebychev("bkg","Chebychev(1)", mB, ROOT.RooArgList(c0, c1))

# Extended yields
nsig = ROOT.RooRealVar("nsig","signal yield",     5.0e4, 0.0, 1e9)
nbkg = ROOT.RooRealVar("nbkg","background yield", 5.0e5, 0.0, 1e10)

# Total model
model = ROOT.RooAddPdf("model","sig+bkg", ROOT.RooArgList(sig, bkg), ROOT.RooArgList(nsig, nbkg))

# Fit
fitres = model.fitTo(ds, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
fitres.Print()

# Plot: data, total, bkg-only, sig-only + legend & parameter box
COL_TOTAL = ROOT.kBlue+1
COL_BKG   = ROOT.kGreen+2
COL_SIG   = ROOT.kRed

c = ROOT.TCanvas("c_fit_all","B_M fit: data + total + bkg + sig", 900, 700)
frame = mB.frame(ROOT.RooFit.Title("B^{+} #rightarrow J/#psi K^{+}  (Signal Crystal Ball + Chebychev 1, with cuts)"), ROOT.RooFit.Bins(60))

# Data
ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))

# Total model
model.plotOn(frame,
             ROOT.RooFit.LineColor(COL_TOTAL),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("model_curve"))

# Background component
model.plotOn(frame,
             ROOT.RooFit.Components("bkg"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(COL_BKG),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("bkg_curve"))

# Signal component
model.plotOn(frame,
             ROOT.RooFit.Components("sig"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(COL_SIG),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}}  [MeV/c^{2}]")
frame.GetYaxis().SetTitle("Events / bin")
frame.Draw()

# Legend
leg = ROOT.TLegend(0.58, 0.68, 0.90, 0.88)
leg.SetBorderSize(1); leg.SetFillColor(0)
leg.SetHeader("Components","C")
if frame.findObject("data_hist"):   leg.AddEntry(frame.findObject("data_hist"),   "Data",                 "lep")
if frame.findObject("model_curve"): leg.AddEntry(frame.findObject("model_curve"), "Total model",          "l")
if frame.findObject("bkg_curve"):   leg.AddEntry(frame.findObject("bkg_curve"),   "Background (Chebychev 1)","l")
sig_label = "Signal (Crystal Ball)" if use_crystal else "Signal (Gaussian)"
if frame.findObject("sig_curve"):   leg.AddEntry(frame.findObject("sig_curve"),   sig_label,              "l")
leg.Draw()

# Parameter summary
box = ROOT.TPaveText(0.14, 0.70, 0.48, 0.88, "NDC")
box.SetFillStyle(0); box.SetBorderSize(0); box.SetTextFont(42); box.SetTextSize(0.035)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
if use_crystal:
    box.AddText(f"#sigma_{{L}} = {sigmaL.getVal():.2f} #pm {sigmaL.getError():.2f} MeV")
    box.AddText(f"#sigma_{{R}} = {sigmaR.getVal():.2f} #pm {sigmaR.getError():.2f} MeV")
else:
    box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.Draw()

c.Draw()
# Save references for analytics
model_cb_cheb1 = model
ds_cb_cheb1 = ds
```

```python
# Compact analytics block, usable for any of the fits below
def quick_fit_analytics(model, ds, obs, nbins=60, name="Fit", sig_range=(5250, 5310)):
    frame = obs.frame(ROOT.RooFit.Bins(nbins))
    ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))
    model.plotOn(frame, ROOT.RooFit.Name("model_curve"))
    model.plotOn(frame, ROOT.RooFit.Components("sig"),
                 ROOT.RooFit.LineStyle(ROOT.kDotted),
                 ROOT.RooFit.LineColor(ROOT.kRed),
                 ROOT.RooFit.Name("sig_curve"))

    # Count floating parameters
    pars = model.getParameters(ds)
    nfloat = sum(1 for p in pars if getattr(p, "isConstant", lambda: False)() is False)

    # Compute chi2/ndf
    chi2ndf = frame.chiSquare("model_curve", "data_hist", nfloat)
    npoints = frame.GetNbinsX()
    ndf = max(1, npoints - nfloat)
    chi2 = chi2ndf * ndf
    aic = 2 * nfloat + 2 * model.createNLL(ds).getVal()
    bic = nfloat * math.log(max(1, ds.numEntries())) + 2 * model.createNLL(ds).getVal()

    print(f"=== {name} ===")
    print(f"χ²/ndf = {chi2ndf:.3f}  |  AIC = {aic:.2f}  |  BIC = {bic:.2f}  |  nfloat = {nfloat}")
    return {"name": name, "chi2/ndf": chi2ndf, "AIC": aic, "BIC": bic, "nfloat": nfloat}
```

```python
# Run it on the Crystal Ball + Chebychev(1) fit
quick_fit_analytics(model_cb_cheb1, ds_cb_cheb1, mB,
                    nbins=60,
                    name="Crystal Ball + Chebychev(1)",
                    sig_range=(5250, 5310))
```

```python
def _count_floats(pdf, data):
    pars = pdf.getParameters(data)
    it = getattr(pars, "createIterator", None) or getattr(pars, "iterator", None)
    it = it()
    n = 0
    obj = it.Next()
    while obj:
        try:
            if obj.isConstant() == False:
                n += 1
        except:
            pass
        obj = it.Next()
    return n

def quick_metrics_all(model, ds, obs, nbins=60, name="model", rng=None):
    """
    Print pull RMS (~1 ideal), binned LRT deviance (smaller is better),
    and AIC/BIC for the fitted RooFit model.
    """
    # (0) Optional range
    if rng is not None:
        obs.setRange("metr", rng[0], rng[1])
        plotrng = ROOT.RooFit.Range("metr")
    else:
        plotrng = ROOT.RooFit.Range(obs.getMin(), obs.getMax())

    # (1) Build a frame & the pulls (RooFit's definition)
    fr = obs.frame(ROOT.RooFit.Bins(nbins), plotrng)
    ds.plotOn(fr, ROOT.RooFit.Name("data_hist"))
    model.plotOn(fr, ROOT.RooFit.Name("model_curve"))
    pull_graph = fr.pullHist("data_hist", "model_curve")

    # Pull RMS (from RooPlot graph)
    pull_rms = float("nan")
    if pull_graph and pull_graph.GetN() > 0:
        vals = []
        xv = ROOT.Double(0.0); yv = ROOT.Double(0.0)
        for i in range(pull_graph.GetN()):
            pull_graph.GetPoint(i, xv, yv)
            vals.append(float(yv))
        # ignore NaNs (can appear in empty bins)
        vals = [v for v in vals if math.isfinite(v)]
        if vals:
            m = sum(vals)/len(vals)
            pull_rms = math.sqrt(sum((v-m)**2 for v in vals)/len(vals))

    # (2) Binned LRT (saturated) deviance
    # Data histogram (counts)
    hD = ROOT.TH1D("hD_tmp", "", nbins, obs.getMin(), obs.getMax())
    hD.Sumw2()
    # Fill from dataset respecting optional range
    for i in range(int(ds.numEntries())):
        a = ds.get(i)[obs.GetName()].getVal()
        if (rng is None) or (rng[0] <= a <= rng[1]):
            hD.Fill(a)

    # Model expected shape → histogram, then scale to expected events
    hM = model.createHistogram("hM_tmp", obs, ROOT.RooFit.Binning(nbins))
    # Expected total events in (optional) range:
    # If extended: use expectedEvents; else use observed counts
    exp_total = None
    try:
        # use named range if set
        if rng is not None:
            exp_total = model.expectedEvents(ROOT.RooArgSet(obs))
            # restrict to range by ratio of integrals
            int_all  = model.createIntegral(ROOT.RooArgSet(obs)).getVal()
            int_rng  = model.createIntegral(ROOT.RooArgSet(obs), ROOT.RooFit.NormSet(ROOT.RooArgSet(obs)),
                                            ROOT.RooFit.Range("metr")).getVal()
            frac = int_rng/int_all if int_all > 0 else 1.0
            exp_total *= frac
        else:
            exp_total = model.expectedEvents(ROOT.RooArgSet(obs))
    except:
        exp_total = hD.Integral()  # fallback if not extended

    # Normalize model hist to expected counts
    sM = hM.Integral()
    if sM > 0:
        hM.Scale(exp_total / sM)
    else:
        # fallback to data normalisation
        sD = hD.Integral()
        if sD > 0:
            hM.Scale(sD)

    # Saturated deviance: 2 * Σ [ n ln(n/μ) - (n-μ) ] (terms with n=0 contribute 2μ)
    dev = 0.0
    n_used = 0
    for b in range(1, nbins+1):
        n = hD.GetBinContent(b)
        mu = hM.GetBinContent(b)
        if mu <= 0: 
            continue
        if n > 0:
            dev += 2.0*(n*math.log(n/mu) - (n-mu))
        else:
            dev += 2.0*mu
        n_used += 1

    # (3) AIC/BIC from the fit NLL
    # Robust NLL creation (extended if available)
    # Work only in range if provided
    if rng is not None:
        nll = model.createNLL(ds, ROOT.RooFit.Range("metr"),
                                   ROOT.RooFit.Extended(True))
    else:
        nll = model.createNLL(ds, ROOT.RooFit.Extended(True))
    nllv = nll.getVal()
    k = _count_floats(model, ds)
    N = int(ds.numEntries())
    AIC = 2*k + 2*nllv
    BIC = k*math.log(max(1, N)) + 2*nllv

    # (4) Print nicely
    tag = f"[{name}]" if name else ""
    print(f"{tag} pull RMS = {pull_rms:.3f}  |  LRT deviance = {dev:.1f} (smaller better)  |  AIC = {AIC:.2f}  |  BIC = {BIC:.2f}  |  k = {k}")

# Example usage, straight after the fit
# quick_metrics_all(model, ds, mB, nbins=60, name="CB + Cheb(2)")
```
