## Initial fit

```python
decay_str = "Hlt2B2CC_BuToJpsiKplus_JpsiToMuMu_Detached/DecayTree"
# 1) Build an RDataFrame over all files (NO extra cuts)
df_all = ROOT.RDataFrame(decay_str, files)

# 2) RooFit observable: mass window for the fit (this is NOT a data filter)
B_M = ROOT.RooRealVar("B_M", "B_M", 5100., 5600.)

# 3) Book a RooDataSet from the dataframe (single column: B_M)
ds_ptr = df_all.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset", "dataset", ROOT.RooArgSet(B_M))),
    ("B_M",)
)
ds = ds_ptr.GetValue()
N = int(ds.sumEntries())
print(f"{N} entries in RooDataSet (within 5100–5600 MeV)")

# ============================
# === Fit model (as before) ==
#   Signal: Gaussian
#   Bkg   : Chebychev(2)
#   Extended fit (nsig, nbkg)
# ============================

# Signal
mean  = ROOT.RooRealVar("mean",  "mean of gaussians", 5275., 5250., 5285.)
sigma = ROOT.RooRealVar("sigma", "width of gaussians",   10.,    5.,   25.)
sig   = ROOT.RooGaussian("sig", "Signal", B_M, mean, sigma)

# Background
c0 = ROOT.RooRealVar("c0", "coefficient #0", -0.30, -1.0, 1.0)
c1 = ROOT.RooRealVar("c1", "coefficient #1",  0.10, -1.0, 1.0)
bkg = ROOT.RooChebychev("bkg", "background p.d.f.", B_M, ROOT.RooArgList(c0, c1))

# Yields (seed with a simple split of total N to help convergence)
nsig = ROOT.RooRealVar("nsig", "number of signal events",     max(0.1*N, 1.), 0., 1e9)
nbkg = ROOT.RooRealVar("nbkg", "number of background events", max(0.9*N, 1.), 0., 1e9)

# Total model = nsig*sig + nbkg*bkg
model = ROOT.RooAddPdf("model", "sig+bkg",
                       ROOT.RooArgList(sig, bkg),
                       ROOT.RooArgList(nsig, nbkg))

# 4) Fit (extended maximum likelihood)
fitres = model.fitTo(ds,
                     ROOT.RooFit.Extended(True),
                     ROOT.RooFit.Save(True),
                     ROOT.RooFit.PrintLevel(-1))
fitres.Print()

print("\n=== Yields from the extended fit ===")
print(f"  nsig = {nsig.getVal():.0f} ± {nsig.getError():.0f}")
print(f"  nbkg = {nbkg.getVal():.0f} ± {nbkg.getError():.0f}")
print(f"  total (fit) ≈ {nsig.getVal()+nbkg.getVal():.0f}   vs dataset entries {N}")

# 5) Plot with legend
c = ROOT.TCanvas("c_fit_nominal", "B_M fit: Gaussian + Cheby", 800, 650)
frame = B_M.frame(ROOT.RooFit.Title("B^{+} \\rightarrow J/#psi K^{+} : Gaussian + Chebychev (no extra cuts)"),
                  ROOT.RooFit.Bins(50))

ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))
model.plotOn(frame, ROOT.RooFit.Name("model_curve"))
model.plotOn(frame, ROOT.RooFit.Components("bkg"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(ROOT.kGreen+2),
             ROOT.RooFit.Name("bkg_curve"))
model.plotOn(frame, ROOT.RooFit.Components("sig"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(ROOT.kRed),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}}  [MeV/c^{2}]")
frame.Draw()

leg = ROOT.TLegend(0.58, 0.68, 0.88, 0.88); leg.SetBorderSize(0); leg.SetFillStyle(0); leg.SetHeader("Components","C")
for name, label, opt in [
    ("data_hist",   "Data",                  "lep"),
    ("model_curve", "Total model",           "l"),
    ("bkg_curve",   "Background (Chebychev)","l"),
    ("sig_curve",   "Signal (Gaussian)",     "l"),
]:
    obj = frame.findObject(name)
    if obj: leg.AddEntry(obj, label, opt)
leg.Draw()

box = ROOT.TPaveText(0.15, 0.72, 0.50, 0.88, "NDC"); box.SetFillStyle(0); box.SetBorderSize(0)
box.SetTextFont(42); box.SetTextSize(0.035)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.Draw()

c.Draw()
```

```python
# -------------------------
# Dataset in [5100,5600] MeV
# -------------------------
mB = ROOT.RooRealVar("B_M", "B_M", 5100., 5600.)
rdf = ROOT.RDataFrame(decay_str, files)  # no extra cuts per your request

# Book a RooDataSet from the dataframe
ds_ptr = rdf.Book(
    ROOT.std.move(ROOT.RooDataSetHelper("dataset","dataset", ROOT.RooArgSet(mB))),
    ("B_M",)
)
ds = ds_ptr.GetValue()
print(f"{int(ds.sumEntries())} entries in dataset (5100–5600 MeV)")

# -------------------------
# Model: Signal (Double-CB or Gaussian) + Chebychev(2), extended
# -------------------------

# Try Double Crystal Ball; if not available, fall back to Gaussian
use_double_cb = hasattr(ROOT, "RooDoubleCB")

if use_double_cb:
    mean  = ROOT.RooRealVar("mean",  "mean",     5279., 5265., 5290.)
    sigma = ROOT.RooRealVar("sigma", "sigma",      10.,    5.,   30.)
    aL    = ROOT.RooRealVar("aL",    "alphaL",     1.5,    0.5,   5.)
    nL    = ROOT.RooRealVar("nL",    "nL",         5.0,    1.5,  50.)
    aR    = ROOT.RooRealVar("aR",    "alphaR",     2.0,    0.5,   5.)
    nR    = ROOT.RooRealVar("nR",    "nR",         5.0,    1.5,  50.)
    sig   = ROOT.RooDoubleCB("sig", "Signal (DoubleCB)", mB, mean, sigma, aL, nL, aR, nR)
else:
    mean  = ROOT.RooRealVar("mean",  "mean",  5279., 5265., 5290.)
    sigma = ROOT.RooRealVar("sigma", "sigma",   10.,    5.,   30.)
    sig   = ROOT.RooGaussian("sig", "Signal (Gauss)", mB, mean, sigma)

# Background: Chebychev(2)
c0 = ROOT.RooRealVar("c0","c0", 0.0, -1.0, 1.0)
c1 = ROOT.RooRealVar("c1","c1", 0.0, -1.0, 1.0)
bkg = ROOT.RooChebychev("bkg","Chebychev(2)", mB, ROOT.RooArgList(c0, c1))

# Extended yields
nsig = ROOT.RooRealVar("nsig","signal yield",     5.0e4, 0.0, 1e9)
nbkg = ROOT.RooRealVar("nbkg","background yield", 5.0e5, 0.0, 1e10)

# Total model
model = ROOT.RooAddPdf("model","sig+bkg", ROOT.RooArgList(sig, bkg), ROOT.RooArgList(nsig, nbkg))

# -------------------------
# Fit
# -------------------------
fitres = model.fitTo(ds, ROOT.RooFit.Extended(True), ROOT.RooFit.Save(True), ROOT.RooFit.PrintLevel(-1))
fitres.Print()

# -------------------------
# Plot: data, total, bkg-only, sig-only + legend & parameter box
# -------------------------
COL_TOTAL = ROOT.kBlue+1
COL_BKG   = ROOT.kGreen+2
COL_SIG   = ROOT.kRed

c = ROOT.TCanvas("c_fit_all","B_M fit: data + total + bkg + sig", 900, 700)
frame = mB.frame(ROOT.RooFit.Title("B^{+} #rightarrow J/#psi K^{+}  (DoubleCB + Chebychev)"), ROOT.RooFit.Bins(60))

# Data
ds.plotOn(frame, ROOT.RooFit.Name("data_hist"))

# Total model
model.plotOn(frame,
             ROOT.RooFit.LineColor(COL_TOTAL),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("model_curve"))

# Background component
model.plotOn(frame,
             ROOT.RooFit.Components("bkg"),
             ROOT.RooFit.LineStyle(ROOT.kDashed),
             ROOT.RooFit.LineColor(COL_BKG),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("bkg_curve"))

# Signal component
model.plotOn(frame,
             ROOT.RooFit.Components("sig"),
             ROOT.RooFit.LineStyle(ROOT.kDotted),
             ROOT.RooFit.LineColor(COL_SIG),
             ROOT.RooFit.LineWidth(3),
             ROOT.RooFit.Name("sig_curve"))

frame.GetXaxis().SetTitle("m_{J/#psi K^{+}}  [MeV/c^{2}]")
frame.GetYaxis().SetTitle("Events / bin")
frame.Draw()

# Legend
leg = ROOT.TLegend(0.58, 0.68, 0.90, 0.88)
leg.SetBorderSize(1); leg.SetFillColor(0)
leg.SetHeader("Components","C")
if frame.findObject("data_hist"):   leg.AddEntry(frame.findObject("data_hist"),   "Data",                 "lep")
if frame.findObject("model_curve"): leg.AddEntry(frame.findObject("model_curve"), "Total model",          "l")
if frame.findObject("bkg_curve"):   leg.AddEntry(frame.findObject("bkg_curve"),   "Background (Chebychev)","l")
sig_label = "Signal (Double CB)" if use_double_cb else "Signal (Gaussian)"
if frame.findObject("sig_curve"):   leg.AddEntry(frame.findObject("sig_curve"),   sig_label,              "l")
leg.Draw()

# Parameter summary
box = ROOT.TPaveText(0.14, 0.70, 0.48, 0.88, "NDC")
box.SetFillStyle(0); box.SetBorderSize(0); box.SetTextFont(42); box.SetTextSize(0.035)
box.AddText(f"nsig = {nsig.getVal():.0f} #pm {nsig.getError():.0f}")
box.AddText(f"nbkg = {nbkg.getVal():.0f} #pm {nbkg.getError():.0f}")
box.AddText(f"mean = {mean.getVal():.2f} #pm {mean.getError():.2f} MeV")
box.AddText(f"#sigma = {sigma.getVal():.2f} #pm {sigma.getError():.2f} MeV")
box.Draw()

c.Draw()
```
